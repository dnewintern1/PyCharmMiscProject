```python
from abc import ABC, abstractmethod
```

```python
class Hominidae(ABC):

    @abstractmethod
    def diet(self):
        pass
    @abstractmethod
    def walk(self):
        pass

    def behaviour(self):
        print('They show complex faical expression and social behaviour')

chimpanzee = Hominidae()
chimpanzee.diet()



class Human(Hominidae):

    def diet(self):
        print('Humans are Omnivorous.')

    def walk(self):
        print('Humans are Omnivorous.')
```

```python
hu = Human()

#example of decorators

class Employee(object):

    def __init__(self, employee_id):

        self.__employee_id = employee_id

    @property
    def employee_id(self):
        return self.__employee_id

    @employee_id.setter
    def employee_id(self, employee_id):
        self.__employee_id = employee_id

    @employee_id.deleter
    def employee_id(self):
        del self.__employee_id


```

```python
emp = Employee('Eddie')
emp.employee_id('radhe')

```
