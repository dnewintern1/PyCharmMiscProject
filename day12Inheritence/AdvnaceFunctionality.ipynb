```python
#special method

class Competition:

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def __repr__(self):#this one will be called when you write the archery as a var
        return "('{}' , {})".format(self.name, self.score)

    def __str__(self): #this one will be called when the print is called
        return "('{}'-{})".format(self.name, self.score)


archery = Competition("Archery", 100)
# repr(archery)
archery
```

```python
#sub special method
class Saving:
    def __init__(self, amount):
        self.amount = amount
        # self.other = other

    def __sub__(self, other):
        return self.amount - other.amount

    def __floordiv__(self, other):
        return self.amount // other.amount

    def __mod__(self, other):
        return self.amount % other.amount

    def __pow__(self, other):
        return self.amount ** other.amount

s1 = Saving(4)
s2 = Saving(4)

s1 ** s2
s2 ** s1


# s1 * s2
```

```python
class Participants:
    def __init__(self):
        self.participants = []
        self.__index = 0

    def add_participant(self,participant):
        self.participants.append(participant)

    def __len__(self):
        return len(self.participants)

    def __iter__(self):
        self.__index = 0
        return self

    def __next__(self):
        if self.__index == len(self.participants):
            raise StopIteration

        p = self.participants[self.__index]
        self.__index += 1
        return p

```

```python
p = Participants()

p.add_participant("radhe")
p.add_participant("karan")
p.add_participant("Mandeep")
p.add_participant("Ishika")
p.add_participant("harsh")

# for i in p.participants:
#     print(i)
```
