```python
from adodbapi.examples.db_print import kw_args


# /Iterating over keyword arguments

def print_details(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

print_details(name="radhe", age=25,surname = "maurya", college ="ZHDCE")
```

```python
# Accessing specific keys safely

def greet(**kwargs):
    name = kwargs.get("name", "Guest")
    print("Hello", name)

greet(name="radhe", guest = 'atul')
```

```python
# Mixing fixed params + **kwargs

def create_user(role, **details):
    print(role, details)

create_user('Analyst', backend = 25, frontend = 'James')

```

```python
def calculate(*nums, **kwargs):

    operation = kwargs.get("operation", "add")  # default add

    if operation == "mul":
        result = 1
        for n in nums:
            result *= n
    else:
        result = 0
        for n in nums:
            result += n

    return result

calculate(1, 2, 4, operation="mul")
```
