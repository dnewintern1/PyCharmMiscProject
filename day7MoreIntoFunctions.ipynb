```python
#defining the global varable inside the function so the right varable are accessed inside the function
city = "bihar"
country = "patna"
def some_new_fn():

    global city,country

    city = city + "-California"
    country = country + "-World"

    print("country :", country)
    print("city : ", city)


some_new_fn()

```
