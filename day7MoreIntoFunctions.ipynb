```python
#pass by reference -- you acan actually change the value inside a function which will also be reflected out side the function


fruits_list = ['Apple','Grapes','Mango','Bananas']

def sort_items_and_remove_last(my_list):
    my_list.sort()
    my_list.pop()

    print('Inside the function: ', my_list)

sort_items_and_remove_last(fruits_list)
fruits_list
```

```python
import math
def area_circle_fn(radius):
    return math.pi * radius **2

def perimeter_circle_fn(radius):
    return 2 * math.pi * radius

def diameter_circle_fn(radius):
    return 2 * radius

def calculate_for_circle(radius,fn):
    return fn(radius)

calculate_for_circle(32,diameter_circle_fn)
```

```python
def calculate(*args,fn):
    return fn(*args)

calculate(10,fn=diameter_circle_fn)
```

```python
def area_rectangle_fn(length,breadth):
    return length * breadth

calculate(20,30,fn=area_rectangle_fn)
```

```python
def add(a,b):
    return a + b
def sub(a,b):
    return a - b
def mul(a,b):
    return a * b
def div(a,b):
    return a / b



def get_functions(operator="+"):

    if operator == "+":
        return add
    if operator == "-":
        return sub
    if operator == "*":
        return mul

    if operator == "/":
        return div


func = get_functions("-")

func(1,2)

cac_dictionary = {
    '+' : add,
    '-' : sub,
    '*' : mul,
    '/' : div
}

func = cac_dictionary['/']
func
```
