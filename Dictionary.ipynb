```python
#program for slicing a list in half

listnummm = ['12' , '13','14','15','16','17','18','19','20']
len_ofthelist = len(listnummm)
print('the lenght of the list before slicing : ', len_ofthelist)
print('the actual list before  : ', listnummm)

half = int(len_ofthelist / 2)

listnum1 = listnummm[half:]
listnum2 = listnummm[:half]

print('the first half', listnum1,len(listnum1))

print('the second half', listnum2,len(listnum2))
```

```python
#making a calculator

numbertobe = float(input('enter a no. to be operated : '))
numbertobe2 = float(input('enter a no. to be operated on second: '))
operator = input('enter a operator: ')

if operator == '+':
    print(numbertobe + numbertobe2)

elif operator == '-':
    print(numbertobe - numbertobe2)
elif operator == '*':
    print(numbertobe * numbertobe2)

elif operator == '/':
    print(numbertobe / numbertobe2)

elif operator == '%':

    print(numbertobe // numbertobe2)

else:
    print('the operator is invalid')
```
