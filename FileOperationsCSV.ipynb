```python
#its not just csv file but alos list

nums = [[10,20,30],
       [40,50,60],
       [90,60,40]]

file = open("testfolder/number.csv", 'w')

with file:
    write = csv.writer(file)
    write.writerows(nums) #inorder to write functions in list in csv files

    
```

```python
#dealing the csv and dictionaries -- dictionaries into csv files
file = open("testfolder/names.csv", 'w')

with file:
    fnames = ['FirstName', 'LastName']
    writer = csv.DictWriter(file,fieldnames=fnames)

    writer.writeheader()
    writer.writerow({"FirstName" : 'sofiea','LastName':'Reyes',})
    writer.writerow({"FirstName" : 'Jerome','LastName':'Jackson',})
    writer.writerow({"FirstName" : 'Jia','LastName':'Zhong',})
```

```python
#CSV Dilects
csv.register_dialect('pipe', delimiter='\t')

with open('testfolder/record_pipe.csv' , 'r') as file:
    reader = csv.reader(file, dialect = 'tab')

    for row in reader:
        print(row)
```

```python
#CSV Dilects are interchangable in python # define your own dilect



csv.register_dialect('plus', delimiter='+', lineterminator= '\n\n\r')

names = [['FirstName' , 'LastName'],
        ['Sofiea','Reyes'],
        ['Jerome','Jackson'],
        ["jia", 'Zhong']]

file =  open('testfolder/record_pipe.csv' , 'w') 

with file:
    file_writer = csv.writer(file, dialect = 'plus')

    for row in names:
        file_writer.writerow(row)
```
