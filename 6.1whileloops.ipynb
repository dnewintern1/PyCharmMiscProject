```python
even_list = []
odd_list = []
number = 0

while number < 50:

    if number % 2 == 0:
        even_list.append(number)

    else:
        odd_list.append(number)


    number = number + 1

print(even_list,odd_list)
```

```python
#program for converting the celicus to farehite

c_degrees = [-20,-25,-10,23,0,5,4,12,3,45,34,56,40]
index = 0

print( "'     C     F")

while index < len(c_degrees):
    C = c_degrees[index]
    F = (9.0/5) * C + 32

    print('%5d %5.lf' % (C,F))

    index += 1
```
