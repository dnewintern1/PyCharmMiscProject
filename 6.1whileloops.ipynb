```python
clean_haystack = ['hay','hay','hay','hay','hay','hey']

unclean_haystack = ['hay','hay','hay','nodt' ,'needle','needle',"hey"]

item = 'needle'

maximum = len(clean_haystack)

i = 0

while i < maximum:

    print(i)

    if item == unclean_haystack[i]:
        print(i,unclean_haystack[i])

        del unclean_haystack[i]
        print(unclean_haystack)
        # break

    i = i+1

else:
    print("you are out of the loop")

print(unclean_haystack)
```

```python
maxnum = int(input("give me a no. : "))
num = 2
print("prime numbers upto :" , maxnum)

while num <= maxnum:
    isPrime = True
    counter = 2

    while counter < num:   # FIX 1
        if num % counter == 0:  # FIX 2
            isPrime = False
            break
        counter += 1

    if isPrime:
        print(num)

    num += 1
```
