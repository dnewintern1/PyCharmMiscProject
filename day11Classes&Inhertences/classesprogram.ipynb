```python
class Student:
    def __init__(self, name, GPA):
        self.__name = name

        self.__GPA = GPA
        self.__clubs = set()
        self.__active = True

        # Getters
    def get_name(self):
        return self.__name

    def get_GPA(self):
        return self.__GPA

    def get_clubs(self):
        return self.__clubs

    def is_active(self):
        return self.__active

    # Setters
    def set_name(self, name):
        self.__name = name

    def set_GPA(self, GPA):
        self.__GPA = GPA

    def add_clubs(self, clubs):
        self.__clubs.add(clubs)

    def remove_clubs(self, clubs):
        self.__clubs.remove(clubs)

    def set_active(self, active):
        self.__active = True

    # Print all details
    def print_details(self):
        print(f"Name: {self.__name}")
        print(f"GPA: {self.__GPA}")
        print(f"Clubs: {self.__clubs}")
        print(f"Active: {self.__active}")

```

```python
#programming using classes

student_details_list = [
    {'name': 'james', 'GPA': 3.5 , 'clubs': ['Yoga', 'yoga']},
    {'name': 'Emily', 'GPA': 3.4 , 'clubs': ['Tennis'], 'active': False},
    {'name': 'MIchael', 'GPA': 3.2 , 'clubs': ['Football', 'chess']},
    {'name': 'joe', 'GPA': 3.5 , 'is_honours_student': True}
]

#to create a list of the student objects
#to create a list of studnet objects
```

```python
def get_students(student_details_list):
    students_list = []

    for student_details in student_details_list:

        if 'name' not in student_details or 'GPA' not in student_details:
            continue

        s = Student(student_details['name'], student_details['GPA'])

        if 'clubs' in student_details:
            for club in student_details['clubs']:
                s.add_clubs(club)

        if 'active' in student_details:
            s.set_active(student_details['active'])

        students_list.append(s)

    return students_list

```

```python
s1 = get_students(student_details_list)
```

```python
print(s1)
```

```python
for student in s1:
    student.print_details()
```
