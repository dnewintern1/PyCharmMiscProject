```python
#importance of data in AI Development

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn
import seaborn.datasets
import tensorflow as tf
import tensorflow_datasets as tfds


digits = sklearn.datasets.load_digits()

#handwritten digit dataset Visualization

figure = plt.figure(figsize=(12,6)) #matplotlib figure

for i in range(1,11):

    image = np.array(digits['images'][i], dtype= 'float') #convert to NumpPy format for visulization

    figure.add_subplot(2,5,i) #adding subplot to matplot lib figure
    plt.inshow(image, cmap ='Greys') #visualize greyscale image

plt.show()
```
