```python
def find_max_in_list(somelistt):

    if len(somelistt) == 0:
        print('zero element in the list')
        return


    max_element  = somelistt[0]

    length = len(somelistt)

    for i in range(length):
        if somelistt[i] > max_element:
            max_element = somelistt[i]

    return max_element

find_max_in_list([1,2,3,4,5])
```

```python
#automatic unpacking of the results

def add_sub(x,y):
    add_result = x + y
    sub_result = x - y

    return add_result,sub_result

result1,resu2 = add_sub(2,3)
print(resu2)
```

```python
def find_first_capital_letter(some_string):
    capital_letter = None

    for ch in some_string:
        if ch.upper() == ch and ch !=" ":
            capital_letter = ch
            break

    if capital_letter is None:
        return 'there is no cap letter in the string'
    else:
        return'the First capital letter  ' + capital_letter
```

```python
find_first_capital_letter("the first Capital letter in this string is")
```

```python
def generate_list(name,num_elements):

    print('generate list using list comprehension')
    returnlist = [name for i in range(num_elements)]

    return returnlist
```
