```python
def find_max_in_list(somelistt):

    if len(somelistt) == 0:
        print('zero element in the list')
        return


    max_element  = somelistt[0]

    length = len(somelistt)

    for i in range(length):
        if somelistt[i] > max_element:
            max_element = somelistt[i]

    return max_element

find_max_in_list([1,2,3,4,5])
```

```python
def find_first_capital_letter(some_string):
    capital_letter = None

    for ch in some_string:
        if ch.upper() == ch and ch !=" ":
            capital_letter = ch
            break

    if capital_letter is None:
        return 'there is no cap letter in the string'
    else:
        return'the First capital letter  ' + capital_letter
```

```python
def maxlist(longlist):

    max = longlist[0]
    for item in longlist:
        if item > max:
            max = item

    return max

maxlist([1,2,3,4,5,34,2,3,5,3,4, 45, 4, 34, 5, 4, 454, 4, 5])




```
