```python
#forming the right angled triangle

for i in range(1,9):

    for j in range(1,i):
        print(j , end="")
    print()

for i in range(1,9):

    for j in range(1,i):
        print("*", end="")
    print()
```

```python
# reverse pyramid

row = 8
for i in range(1,row+1):
    print(' ' *(i), end='')

    for j in range(1,row-i+1):
        print(' ', end='')
        print(j , end="")

    print()
```

```python
# reverse star piryamid

row = 8
for i in range(1,row+1):
    print(' ' *(i), end='')

    for j in range(1,row-i+1):
        print(' ', end='')
        print("*" , end="")

    print()
```
