```python
def check_prime1(number):

    num_iteration =0

    for i in range(2,number):
        num_iteration+=1

        if number % i == 0:
            print("%d is not a prime number \nTotal number of iterations = %d" %(number,num_iteration))
            return

    print('%d is a prime number \nTotal number of iterations = %d' %(number,num_iteration))
```

```python
check_prime1(193)

check_prime1(1847)
```

```python
def check_prime2(number):

    num_iteration =0

    mid_point = int(number/2)

    for i in range(2,mid_point):
        num_iteration+=1

        if number % i == 0:
            print("%d is not a prime number \nTotal number of iterations = %d" %(number,num_iteration))
            return

    print('%d is a prime number \nTotal number of iterations = %d' %(number,num_iteration))
```

```python
check_prime2(193)
```

```python
check_prime2(1847)
```

```python
def find_maxval(num_list):

    maxval = num_list[0]
    num_iteration = 0

    for i in range(len(num_list)):
        num_iteration += 1

        if num_list[i] > maxval:
            maxval = num_list[i]

    print("Max value is: %d ; the no. of itterations %d" % (maxval,num_iteration))
```

```python
find_maxval([12,13,14,15,16,16,16,17,17,3])
```

```python
def find_factorial(number:int):
    num_iteration =0
    fact = 1

    if number <1 or type(number) != int:
        print("The input is not a positive integer")
        return

    for i in range(1,number+1):
        fact = fact * i
        num_iteration += 1

    print("The factorial of ",fact, "No. of itteration" , num_iteration)
```

```python
find_factorial(10)
```

```python
find_factorial(1)
```

```python
find_factorial(1050)
```

```python
def print_pairs(number_list):

    num_iteration = 0
    n = len(number_list)

    for i in range(n):
        for j in range(n):
            print(number_list[i] , number_list[j], end="\n")
            num_iteration += 1

    print("total iterations = %d" %num_iteration)
```

```python
print_pairs([[1,2,3],[4,5,6]])
```
