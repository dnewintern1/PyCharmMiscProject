```python
def bubble_sort(arr):
    for n in range(len(arr) - 1, 0, -1):
        swapped = False
        for i in range(n):
            if arr[i] > arr[i + 1]:
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
                swapped = True
        if not swapped:
            break

# Example:
arr = [39, 12, 18, 85, 72, 10, 2, 18]
bubble_sort(arr)
print("Sorted:", arr)
```
