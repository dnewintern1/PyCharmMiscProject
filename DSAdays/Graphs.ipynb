```python
#getting started with graph


#binary search

def binary_serach(sorted_list,key):
    min_index = 0
    max_index = len(sorted_list)-1
    print(max_index)

    while min_index <= max_index:
        mid = (min_index + max_index)//2
        print(mid)

        if sorted_list[mid] == key:
            print("block rann ====")
            return mid

        elif sorted_list[mid] < key:
            print("?>>>>>>>>>>>>>>")
            min_index = mid + 1
            return "element found" , sorted_list[min_index]

        else:
            print("<<<<<<<<<<else block>>>>>>>>>>")
            max_index = mid - 1

    return -1

```

```python
num_list = [1,2,3,4,5,6,7,8,9]

binary_serach(num_list,1)
```
