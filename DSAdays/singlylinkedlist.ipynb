```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None

a = Node(12)
b = Node(13)
c = Node(14)
d = Node(15)


a.next = b
b.next = c
c.next = d


def interation_step(head):
    curr = head
    # print(curr.data)

    while curr != None:
        print(curr.data, end="--->")
        curr = curr.next

head = a

interation_step(head)

#addition at Fronnt




```

```python
#additon at the front of the node

e = Node(1)
e.next = a
head = e
# print(head.data)

interation_step(head)
```

```python
# addition at the end of the node

# print(head.data) #head is at e = 1
head = e
g = Node(99)
g.next = None
curr = head

while curr.next != None:
    curr = curr.next
    # print(curr)

curr.next = g

interation_step(head)
```
